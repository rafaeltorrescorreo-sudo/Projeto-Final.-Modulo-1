# Fase 1: Análise Exploratória (EDA)
Nesta fase inicial, importamos as bibliotecas necessárias, carregamos os dados e realizamos uma inspeção inicial das dimensões, tipos de dados e estatísticas descritivas. Também plotamos gráficos analíticos para entender a distribuição dos dados e o desbalanceamento da variável alvo.

```python
# Importação das bibliotecas
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuração de estilo para os gráficos
sns.set_theme(style="whitegrid")

# Carregar o dataset
# URL direta
url = "https://raw.githubusercontent.com/rafaeltorrescorreo-sudo/Projeto-Final.-Modulo-1/main/data/manutencao_preditiva.csv"

# Cargar arquivo
df = pd.read_csv(url)

# Apresentar dimensões e tipos de variáveis
print(f"Dimensões do dataset: {df.shape[0]} linhas e {df.shape[1]} colunas.\n")
print("Tipos de dados:")
print(df.dtypes)

# Resumo estatístico descritivo
display(df.describe())
```

```python
# Gráfico 1: Histograma de distribuição de uma variável preditora
plt.figure(figsize=(8, 5))
sns.histplot(df['velocidade_rotacao_rpm'], bins=30, kde=True, color='blue')
plt.title('Distribuição da Velocidade de Rotação (RPM)')
plt.xlabel('Velocidade de Rotação (RPM)')
plt.ylabel('Frequência')
plt.show()

# Gráfico 2: Gráfico de barras comprovando o desbalanceamento da variável alvo
plt.figure(figsize=(6, 4))
sns.countplot(x='falha_maquina', data=df, palette='Set1')
plt.title('Distribuição da Variável Alvo (Falha na Máquina)')
plt.xlabel('Status (0 = Normal, 1 = Falha)')
plt.ylabel('Contagem')
plt.show()

# Gráfico 3: Mapa de calor (Heatmap) com a correlação de Pearson
plt.figure(figsize=(10, 8))
corr = df.select_dtypes(include=[np.number]).corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Matriz de Correlação de Pearson')
plt.show()
```

**Análise da Fase 1:**

Observando o resumo estatístico e os gráficos, notamos que a variável alvo `Falha` é altamente desbalanceada (a quantidade de classe 0 é muito superior à classe 1). Isso direciona nossa estratégia para a obrigatoriedade do uso de técnicas de reamostragem (como SMOTE) no futuro para evitar que o modelo tenda a prever apenas a classe majoritária. Além disso, a matriz de correlação mostra quais variáveis têm maior colinearidade, e os histogramas revelam a escala divergente entre as features numéricas, o que exigirá normalização (StandardScaler) dependendo do algoritmo utilizado.

# Fase 2: Limpeza e Tratamento de Dados (Data Prep)
Realizamos a remoção de registros duplicados e o tratamento de valores nulos. Optou-se pela imputação via Mediana devido à provável presença de outliers, que distorcem a Média.

```python
# Identificar e remover linhas duplicadas
duplicatas_antes = df.duplicated().sum()
df = df.drop_duplicates()
print(f"Linhas duplicadas removidas: {duplicatas_antes}")

# Tratamento de nulos: Imputação pela Mediana
# A mediana foi escolhida em vez da média por ser uma medida de tendência central robusta a outliers.
# Como os dados de maquinário frequentemente apresentam picos de anomalias (outliers), a média seria distorcida.
colunas_numericas = df.select_dtypes(include=[np.number]).columns
df[colunas_numericas] = df[colunas_numericas].fillna(df[colunas_numericas].median())
print("Valores nulos após imputação:\n", df.isnull().sum())

# Gráficos Boxplot para identificar outliers nas variáveis explicativas
plt.figure(figsize=(12, 6))
sns.boxplot(data=df[['velocidade_rotacao_rpm', 'torque_nm']])
plt.title('Boxplot das Variáveis Preditoras (Detecção de Outliers)')
plt.show()
```

# Fase 3: Feature Engineering
Criação de uma nova variável numérica com base em variáveis existentes para ajudar o modelo a capturar relacionamentos mais complexos.

```python
# Criação da variável 'potencia'.
# A lógica de negócio dita que a potência mecânica é proporcional ao torque multiplicado pela rotação.
# Como os nulos já foram tratados na Fase 2, esta operação é segura.
df['potencia'] = df['velocidade_rotacao_rpm'] * df['torque_nm']

# Verificar as primeiras linhas para confirmar a criação
display(df[['velocidade_rotacao_rpm', 'torque_nm', 'potencia']].head())
```

# Fase 4: Divisão e Balanceamento dos Dados
Separação entre features (X) e target (y), divisão em treino e teste com estratificação, e aplicação do SMOTE APENAS no conjunto de treino para evitar Data Leakage (vazamento de dados).

```python
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE

# Separar variáveis preditoras (X) da variável alvo (y)
#X = df.drop(columns=['falha_maquina'])
#y = df['falha_maquina']
X = df.select_dtypes(include=[np.number]).drop(columns=['falha_maquina'], errors='ignore')
y = df['falha_maquina']

# Divisão em treino (80%) e teste (20%) mantendo a proporção das classes (stratify=y)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

# Aplicar técnica de reamostragem (SMOTE) SOMENTE no conjunto de treino
smote = SMOTE(random_state=42)
X_train_bal, y_train_bal = smote.fit_resample(X_train, y_train)

print(f"Distribuição do target no treino APÓS SMOTE:\n{y_train_bal.value_counts()}")
```

# Fase 5: Escalonamento de Variáveis (StandardScaler)
O KNN baseia-se em cálculo de distância (Euclidiana, por exemplo). Variáveis com escalas maiores dominariam o modelo, logo o StandardScaler é obrigatório. Já a Árvore de Decisão faz divisões nos dados baseadas em regras lógicas (thresholds), sendo imune à escala dos atributos.

```python
from sklearn.preprocessing import StandardScaler

# Escalonamento apenas para o KNN
scaler = StandardScaler()

# Usamos fit_transform no treino para aprender os parâmetros (média e desvio padrão) e aplicá-los
X_train_knn = scaler.fit_transform(X_train_bal)
# Usamos apenas transform no teste para evitar Data Leakage
X_test_knn = scaler.transform(X_test)

# Para a Árvore de Decisão, mantemos os dados com sua escala original
X_train_tree = X_train_bal.copy()
X_test_tree = X_test.copy()

# --- CÓDIGO DE VERIFICAÇÃO DO ESCALONAMENTO ---

print("=== 1. VERIFICAÇÃO DE DIMENSÕES ===")
print(f"X_train_knn formato:  {X_train_knn.shape}")
print(f"X_test_knn formato:   {X_test_knn.shape}")
print(f"X_train_tree formato: {X_train_tree.shape}")
print(f"X_test_tree formato:  {X_test_tree.shape}\n")

# Calcular média e desvio padrão do treino escalonado
media_treino = np.mean(X_train_knn, axis=0)
std_treino = np.std(X_train_knn, axis=0)

print("=== 2. VERIFICAÇÃO ESTATÍSTICA (KNN TREINO) ===")
# Arredondamos para 4 casas decimais para facilitar a leitura
print(f"Média de cada coluna (deve ser próxima de 0):\n{np.round(media_treino, 4)}")
print(f"\nDesvio Padrão de cada coluna (deve ser igual a 1):\n{np.round(std_treino, 4)}\n")

print("=== 3. COMPARAÇÃO DE VALORES (PRIMEIRA LINHA) ===")
print("Valores originais (para Árvore de Decisão):")
print(X_train_tree.iloc[0].values)
print("\nValores padronizados (para KNN):")
```

# Fase 6: Ajuste de Parâmetros e Combate ao Overfitting
Treinamento dos modelos KNN e Árvore de Decisão variando hiperparâmetros para analisar o comportamento da acurácia e detectar sobreajuste (overfitting).

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

print("--- Avaliação KNN ---")
k_valores = [3, 5, 7]
for k in k_valores:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_knn, y_train_bal)

    acc_treino = accuracy_score(y_train_bal, knn.predict(X_train_knn))
    acc_teste = accuracy_score(y_test, knn.predict(X_test_knn))
    print(f"KNN (K={k}) -> Acurácia Treino: {acc_treino:.4f} | Acurácia Teste: {acc_teste:.4f}")

print("\n--- Avaliação Árvore de Decisão ---")
depth_valores = [3, 5, None]
for depth in depth_valores:
    tree = DecisionTreeClassifier(max_depth=depth, random_state=42)
    tree.fit(X_train_tree, y_train_bal)

    acc_treino = accuracy_score(y_train_bal, tree.predict(X_train_tree))
    acc_teste = accuracy_score(y_test, tree.predict(X_test_tree))
    print(f"Árvore (max_depth={depth}) -> Acurácia Treino: {acc_treino:.4f} | Acurácia Teste: {acc_teste:.4f}")
```

**Análise de Overfitting:**
- Na Árvore de Decisão, ao usar `max_depth=None`, a acurácia de treino chega a 1.0000 (100%), mas a acurácia de teste cai. Isso é um claro sintoma de *overfitting* (o modelo decorou o treino e falha em dados novos). O uso de `max_depth=5` ou `3` atua como uma "poda", limitando o crescimento da árvore e garantindo uma melhor generalização (estabilidade no teste).
- No KNN, valores menores de K (como K=3) tendem a apresentar maior variação. Um K intermediário (como K=5 ou K=7) suaviza as fronteiras de decisão e estabiliza o modelo contra ruídos, melhorando o desempenho no teste.

# Fase 7: Avaliação da Acurácia e Veredito Final
Comparação do melhor KNN contra a melhor Árvore de Decisão para definir qual modelo irá para produção.

```python
# Instanciando os melhores modelos baseados na experimentação acima
best_knn = KNeighborsClassifier(n_neighbors=5)
best_knn.fit(X_train_knn, y_train_bal)
acc_final_knn = accuracy_score(y_test, best_knn.predict(X_test_knn))

best_tree = DecisionTreeClassifier(max_depth=5, random_state=42)
best_tree.fit(X_train_tree, y_train_bal)
acc_final_tree = accuracy_score(y_test, best_tree.predict(X_test_tree))

print("=== VEREDITO FINAL ===")
print(f"Acurácia no Teste - Melhor KNN (K=5): {acc_final_knn:.4f}")
print(f"Acurácia no Teste - Melhor Árvore (max_depth=5): {acc_final_tree:.4f}")

if acc_final_tree > acc_final_knn:
    print("\nO modelo selecionado para implantação pela empresa é a ÁRVORE DE DECISÃO.")
else:
    print("\nO modelo selecionado para implantação pela empresa é o KNN.")
```

**Conclusão Final:**

A escolha técnica não se baseia na performance durante o treino, mas na capacidade de generalização com dados invisíveis (conjunto de teste). O modelo selecionado acima apresentou o melhor balanço numérico de acurácia com os dados não vistos. Na Indústria 4.0, essa estabilidade significa prever falhas reais sem acionar alarmes falsos excessivos (consequência comum em modelos em overfitting). Portanto, ele é a opção tecnicamente mais sólida para evitar paradas na linha de produção.
